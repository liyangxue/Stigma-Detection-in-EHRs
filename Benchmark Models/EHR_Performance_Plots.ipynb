```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# Confusion matrices (fixed order)
cms = [
    ("GPT-4o Few-Shot", np.array([[181, 44], [222, 379]])),
    ("GPT-4o Zero-Shot", np.array([[196, 29], [408, 193]])),
    ("Longformer: SST",  np.array([[175, 50], [ 40, 561]])),
    ("Longformer: T",    np.array([[159, 66], [ 50, 551]])),
]
classes = ['Nonstigma', 'Stigma']

# Figure: exact 6in wide
fig = plt.figure(figsize=(6, 6)) # Increased width to 6 inches
# 2x3 grid, last column reserved for a single shared colorbar
gs = GridSpec(2, 3, figure=fig, width_ratios=[1, 1, 0.06], wspace=0.5, hspace=0.6) # Increased wspace and hspace

# Create axes and a dedicated colorbar axis
axes = [fig.add_subplot(gs[i//2, i%2]) for i in range(4)]
cax = fig.add_subplot(gs[:, 2])

# Shared color scale across all plots
vmax = max(cm.max() for _, cm in cms)
vmin = 0
im_last = None

for i, (ax, (title, cm)) in enumerate(zip(axes, cms)):
    im_last = ax.imshow(cm, cmap='Blues', vmin=vmin, vmax=vmax)
    ax.set_title(title, fontsize=8, pad=3)
    ax.set_aspect('equal')

    # Ticks & labels
    ax.set_xticks(np.arange(len(classes)))
    ax.set_xticklabels(classes, fontsize=6)
    ax.set_yticks(np.arange(len(classes)))
    ax.set_yticklabels(classes, fontsize=6)

    # Cell annotations
    thresh = (vmin + vmax) / 2.0
    for r in range(cm.shape[0]):
        for c in range(cm.shape[1]):
            ax.text(c, r, f"{cm[r, c]:d}",
                    ha="center", va="center",
                    fontsize=8,
                    color="white" if cm[r, c] > thresh else "black")

    # Thinner spines
    for spine in ax.spines.values():
        spine.set_linewidth(0.5)

# Add shared labels
fig.supxlabel('Predicted label', fontsize=8)
fig.supylabel('Actual label', fontsize=8)


# Single shared colorbar (legend)
cb = fig.colorbar(im_last, cax=cax)
cb.ax.tick_params(labelsize=8)
cb.set_label('Count', fontsize=8)

# Save exactly 4"x4" at 1200 dpi
out_path = '/content/drive/My Drive/EHR_PROJ/Figures/Combined_Confusion_Matrices_6x6.eps'
plt.savefig(out_path, format = "eps") # Removed bbox_inches='tight'
plt.show()
print(f"Saved to: {out_path}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec
import pandas as pd

# Confusion matrices (fixed order)
cms_dict = {
    "GPT-4o Few-Shot": np.array([[181, 44], [222, 379]]),
    "GPT-4o Zero-Shot": np.array([[196, 29], [408, 193]]),
    "Longformer: SST": np.array([[167, 58], [27, 574]]),
    "Longformer: T": np.array([[159, 66], [50, 551]]),
}
classes = ['Nonstigma', 'Stigma']

# --- Compute metrics ---
def compute_metrics(cm):
    tn, fp, fn, tp = cm.ravel()
    acc = (tp + tn) / cm.sum()
    prec = tp / (tp + fp) if (tp + fp) > 0 else 0
    rec = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * prec * rec / (prec + rec) if (prec + rec) > 0 else 0
    return acc, prec, rec, f1

metrics = {name: compute_metrics(cm) for name, cm in cms_dict.items()}

# --- Reorder models as requested ---
ordered_models = ["GPT-4o Zero-Shot", "GPT-4o Few-Shot", "Longformer: T", "Longformer: SST"]
ordered_colors = ['#a8ddb5', '#7bccc4', '#4eb3d3', '#0868ac']
metric_names = ["Accuracy", "Precision", "Recall", "F1"]

metric_values = np.array([metrics[m] for m in ordered_models])

# --- Figure layout ---
fig = plt.figure(figsize=(9, 10))
gs = GridSpec(3, 3, figure=fig, width_ratios=[1, 1, 0.06],
              height_ratios=[1, 1, 1.1], wspace=0.5, hspace=0.6)

# Confusion matrices
cms_for_plot = [(name, cms_dict[name]) for name in ordered_models]
axes = [fig.add_subplot(gs[i//2, i%2]) for i in range(4)]
cax = fig.add_subplot(gs[:2, 2])
vmax = max(cm.max() for _, cm in cms_for_plot)
vmin = 0

for ax, (title, cm) in zip(axes, cms_for_plot):
    im = ax.imshow(cm, cmap='Blues', vmin=vmin, vmax=vmax)
    ax.set_title(title, fontsize=9, pad=3)
    ax.set_aspect('equal')
    ax.set_xticks(np.arange(len(classes)))
    ax.set_xticklabels(classes, fontsize=7)
    ax.set_yticks(np.arange(len(classes)))
    ax.set_yticklabels(classes, fontsize=7)

    thresh = (vmin + vmax) / 2.0
    for r in range(cm.shape[0]):
        for c in range(cm.shape[1]):
            ax.text(c, r, f"{cm[r, c]:d}",
                    ha="center", va="center",
                    fontsize=8,
                    color="white" if cm[r, c] > thresh else "black")

    for spine in ax.spines.values():
        spine.set_linewidth(0.5)

fig.supxlabel('Predicted label', fontsize=9)
fig.supylabel('Actual label', fontsize=9)
cb = fig.colorbar(im, cax=cax)
cb.ax.tick_params(labelsize=8)
cb.set_label('Count', fontsize=8)

# --- Performance metrics chart ---
ax_metrics = fig.add_subplot(gs[2, :3])
x = np.arange(len(metric_names))
width = 0.18

for i, (model, color) in enumerate(zip(ordered_models, ordered_colors)):
    bars = ax_metrics.bar(x + (i - 1.5) * width, metric_values[i, :], width,
                          label=model, color=color, edgecolor='black', linewidth=0.3)
    for bar in bars:
        height = bar.get_height()
        ax_metrics.text(bar.get_x() + bar.get_width()/2, height + 0.02,
                        f"{height:.2f}", ha='center', va='bottom', fontsize=8)

ax_metrics.set_xticks(x)
ax_metrics.set_xticklabels(metric_names, fontsize=10)
ax_metrics.set_ylim(0, 1.05)
ax_metrics.set_ylabel("Score", fontsize=9)
ax_metrics.set_title("Performance Metrics Derived from Confusion Matrices", fontsize=10, pad=10)

# Legend placed BELOW chart
ax_metrics.legend(fontsize=9, ncol=4, loc='upper center', bbox_to_anchor=(0.5, -0.15), frameon=False)

# Save & show
out_path = '/content/drive/My Drive/EHR_PROJ/Figures/Confusion_and_Metrics_GPT_vs_Longformer.eps'
plt.savefig(out_path, format="eps", bbox_inches='tight')
plt.show()
print(f"Saved to: {out_path}")

# --- Display table ---
df_metrics = pd.DataFrame(metric_values, columns=metric_names, index=ordered_models)
print(df_metrics.round(3))
```
